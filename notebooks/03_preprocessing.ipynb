```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from pathlib import Path
```

```python
ROOT_DIR = Path.cwd().parent

DATA_PATH = ROOT_DIR / "data" / "processed" / "churn_clean.csv"

save_path = ROOT_DIR / "notebooks" / "churn_preprocessed.csv"

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
df.drop(columns=['customerID'], inplace=True)
```

```python
df.head(3)
```

```python
# 3. Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
print('Non-numeric or blank TotalCharges:', df["TotalCharges"].isnull().sum())
```

```python
df[df["TotalCharges"].isnull()]
```

```python
# 4. Handle missing TotalCharges
df["TotalCharges"] = df["TotalCharges"].fillna(df["tenure"] * df["MonthlyCharges"])
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
df.head(3)
```

```python
print(df[['Churn', 'Partner', 'Dependents',
          'PhoneService', 'gender']].isna().sum())
```

```python
df.to_csv(
    save_path,
    index=False
)
```
